# Anomaly
## DBSCAN for Anomalies 

```python
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import pandas as pd
```

```python
X, y = make_moons(
    n_samples=500,
    noise=0.1,
    random_state=42
)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled.shape
```

```python
# Visualize
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1])
```

```python
#DBSCAN
dbscan=DBSCAN(eps=0.18,
              min_samples=5)
labels=dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],c=labels)
```

## Isolation Forest

```python
df = pd.read_csv("thyroid_dataset.csv")
```

```python
df.head()
```

```python
X = df.drop("Outlier_label", axis=1)
y = df["Outlier_label"]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
from sklearn.ensemble import IsolationForest

clf = IsolationForest(n_estimators=200, contamination=0.036, random_state=42)
labels = clf.fit_predict(X_scaled)
```

```python
labels
```

```python
# Visualize
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
```

```python
import numpy as np

n_outliers = np.sum(labels == -1)
n_normal = np.sum(labels == 1)

print("outlier = ", n_outliers)
print("normal = ", n_normal)
```

## LOF - >Local Outlier Factor

```python
from sklearn.neighbors import LocalOutlierFactor

neighbs = LocalOutlierFactor(contamination=0.036)

labels = neighbs.fit_predict(X_scaled)

```

```python
plt.figure(figsize=(8, 6))

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
```

```python
n_outliers = np.sum(labels == -1)
n_normal = np.sum(labels == 1)

print("outlier = ", n_outliers)
print("normal = ", n_normal)
```
