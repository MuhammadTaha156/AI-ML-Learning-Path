#  UnSupervised ML (Part-1)
##    K - Mean Clustering

```python
import seaborn as sns
from sklearn.datasets import make_blobs
import pandas as pd
```

```python
X,y=make_blobs(
    n_samples=500,
    n_features=2,
    centers=4,
    random_state=42
)
```

```python
y
```

```python
# Visualize
sns.scatterplot(x=X[:,0],y=X[:,1])
```

```python
# K - Means
from sklearn.cluster import KMeans
```

```python
K=4
k_model=KMeans(n_clusters=K,random_state=42)
```

```python
labels=k_model.fit_predict(X)
#Label = Cluster Number
```

```python
labels
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1],c=labels)
```

## Choose our K Value - Elbow, Silhousette Score
### Elbow Method

```python
wcss=[]
for k in range(1,21):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    print(kmeans.inertia_)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
sns.lineplot(x=range(1,21),y=wcss,marker="o")
```

```python
# Kneed Module
!pip install kneed
```

```python
from kneed import KneeLocator
```

```python
knee=KneeLocator(range(1,21),wcss,curve="convex",direction="decreasing")
```

```python
print("Optimal K = ",knee.elbow)
```

# Sihouette Score

```python
from sklearn.metrics import silhouette_score
```

```python
ss=[]

for k in range(2,21):
    kmeans=KMeans(n_clusters=k)
    labels=kmeans.fit_predict(X)
    score=silhouette_score(X,labels)

    ss.append(score)
    
ss
```

```python
#  Plot - K and ss
sns.lineplot(x=range(2,21),y=ss,marker="o")
```

# K - Means for Iris Dataset

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
X=iris.data
y=iris.target
```

```python
X
```

```python
# Visualize
sns.scatterplot(x=X[:,0],y=X[:,1],c=y)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()

X_scaled=scaler.fit_transform(X)
```

```python
# Optional - Dimensionally Reduction using - PCA
from sklearn.decomposition import PCA

pca=PCA(n_components=2)
pca_Data=pca.fit_transform(X_scaled)
```

```python

# Elbow Method

wcss=[]

for k in range(1,11):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit_predict(pca_Data)
    wcss.append(kmeans.inertia_)
    
```

```python
wcss
```

```python
sns.lineplot(x=range(1,11),y=wcss,marker="o")
```

```python
# KMeans
kmeans=KMeans(n_clusters=3,random_state=10)
labels=kmeans.fit_predict(pca_Data)

sns.scatterplot(x=pca_Data[:,0],y=pca_Data[:,1],c=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,1],marker="X",c="red",s=200)
```
